1.data prepartion

```python
import os, glob, hashlib
import tensorflow as tf
import pandas as pd
from sklearn.model_selection import train_test_split

ROOT = "/Users/fuyuhan/Desktop/2750"   # directory: ROOT/class_name/*.jpg
IMG_SIZE = 64
BATCH = 32
SEED = 42

# 1) scan the file and build the DataFrame
rows = []
for cls in sorted(os.listdir(ROOT)):
    p = os.path.join(ROOT, cls)
    if not os.path.isdir(p): continue
    for f in glob.glob(os.path.join(p, "*.jpg")) + glob.glob(os.path.join(p, "*.png")):
        rows.append({"path": f, "label": cls})
df = pd.DataFrame(rows)

# ：remove the file which is not valid RGB image
def is_valid_rgb(path):
    try:
        img = tf.io.read_file(path)
        img = tf.io.decode_image(img, channels=3, expand_animations=False)
        return img.shape[-1] == 3
    except Exception:
        return False

df = df[df["path"].apply(is_valid_rgb)].reset_index(drop=True)

# deduplication based on file hash
def file_hash(p, n=2<<20):  # 2MB
    h = hashlib.md5()
    with open(p, "rb") as f:
        for chunk in iter(lambda: f.read(n), b""):
            h.update(chunk)
    return h.hexdigest()
df["hash"] = df["path"].apply(file_hash)
df = df.drop_duplicates(subset=["hash"]).drop(columns=["hash"]).reset_index(drop=True)

# 2) split train/val/test
train_df, temp_df = train_test_split(
    df, test_size=0.30, stratify=df["label"], random_state=SEED
)
val_df, test_df = train_test_split(
    temp_df, test_size=0.50, stratify=temp_df["label"], random_state=SEED
)  # train/val/test 70/15/15

class_names = sorted(df["label"].unique())
class_to_id = {c:i for i,c in enumerate(class_names)}
for d in (train_df, val_df, test_df):
    d["y"] = d["label"].map(class_to_id)


AUTO = tf.data.AUTOTUNE

def decode_resize_norm(path, label):
    img = tf.io.read_file(path)
    img = tf.io.decode_image(img, channels=3, expand_animations=False)
    img = tf.image.resize(img, (IMG_SIZE, IMG_SIZE))
    img = tf.cast(img, tf.float32) / 255.0
    return img, tf.cast(label, tf.int32)

data_augment = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"),
    tf.keras.layers.RandomRotation(0.05),
    tf.keras.layers.RandomZoom(0.1),
    tf.keras.layers.RandomContrast(0.1),
])

def make_ds(df, training=False):
    ds = tf.data.Dataset.from_tensor_slices((df["path"].values, df["y"].values))
    ds = ds.map(decode_resize_norm, num_parallel_calls=AUTO)
    if training:
        ds = ds.shuffle(2048, seed=SEED, reshuffle_each_iteration=True)
        ds = ds.map(lambda x,y: (data_augment(x, training=True), y), num_parallel_calls=AUTO)
    ds = ds.batch(BATCH).prefetch(AUTO)
    return ds

train_ds = make_ds(train_df, training=True)
val_ds   = make_ds(val_df, training=False)
test_ds  = make_ds(test_df, training=False)

# （op）calculate class weights
import numpy as np
counts = train_df["y"].value_counts().sort_index().to_numpy()
class_weights = {i: (len(train_df)/ (len(class_names)*counts[i])) for i in range(len(class_names))}

#  Keras train：
# model.fit(train_ds, validation_data=val_ds, epochs=..., class_weight=class_weights)
```

3.model selection


```python
import tensorflow as tf
from tensorflow.keras import layers, models

NUM_CLASSES = len(class_names)  # 10 classes：Annual Crop, Forest ...
IMG_SIZE = 64                   # size: 64x64

def res_block(x, filters, stride=1):
    """
    1 ResNet Basic Block：
    - two 3x3 Conv + BN + ReLU
    - （shortcut）
    """
    shortcut = x

    # the first convolution： 
    x = layers.Conv2D(filters, kernel_size=3, strides=stride,
                      padding="same", use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)

    # the second convolution：
    x = layers.Conv2D(filters, kernel_size=3, strides=1,
                      padding="same", use_bias=False)(x)
    x = layers.BatchNormalization()(x)

    # if channel or size changed，use 1x1 conv to adjust shortcut
    if stride != 1 or shortcut.shape[-1] != filters:
        shortcut = layers.Conv2D(filters, kernel_size=1, strides=stride,
                                 padding="same", use_bias=False)(shortcut)
        shortcut = layers.BatchNormalization()(shortcut)

    #   ReLU
    x = layers.Add()([x, shortcut])
    x = layers.ReLU()(x)
    return x


def make_resnet18(input_shape=(64, 64, 3), num_classes=10):
    """
    
    - Conv7x7 + MaxPool
    - 4 stage，for each stage ,2  BasicBlock
      filters: 64, 128, 256, 512
    """
    inputs = layers.Input(shape=input_shape)

    # stem
    x = layers.Conv2D(64, kernel_size=7, strides=2,
                      padding="same", use_bias=False)(inputs)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)
    x = layers.MaxPool2D(pool_size=3, strides=2, padding="same")(x)

    # stage 1：64 filters, 2 blocks
    x = res_block(x, 64, stride=1)
    x = res_block(x, 64, stride=1)

    # stage 2：128 filters, 2 blocks（the first bloock stride=2 sampling）
    x = res_block(x, 128, stride=2)
    x = res_block(x, 128, stride=1)

    # stage 3：256 filters, 2 blocks
    x = res_block(x, 256, stride=2)
    x = res_block(x, 256, stride=1)

    # stage 4：512 filters, 2 blocks
    x = res_block(x, 512, stride=2)
    x = res_block(x, 512, stride=1)

    # pooling and output
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.4)(x)  # larger dropout to prevent overfitting
    outputs = layers.Dense(num_classes, activation="softmax")(x)

    model = models.Model(inputs, outputs, name="ResNet18_EuroSAT")
    return model
```

training

```python
import tensorflow as tf


tf.config.optimizer.set_experimental_options(
    {"disable_meta_optimizer": True}
)

# use legacy version Ada for mac M1/M2
optimizer = tf.keras.optimizers.legacy.Adam(learning_rate=1e-3)

resnet18.compile(
    optimizer=optimizer,
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)
```

```python
import tensorflow as tf

# 1️ close meta optimizer，avoid model_pruner crash
tf.config.optimizer.set_experimental_options(
    {"disable_meta_optimizer": True}
)

# 2️ avoid GPU usage, force CPU only
with tf.device("/CPU:0"):
    resnet18 = make_resnet18(
        input_shape=(IMG_SIZE, IMG_SIZE, 3),
        num_classes=NUM_CLASSES
    )

    optimizer = tf.keras.optimizers.legacy.Adam(learning_rate=1e-3)

    resnet18.compile(
        optimizer=optimizer,
        loss="sparse_categorical_crossentropy",  #  y is 0~9 integer
        metrics=["accuracy"],
    )

    callbacks = [
        tf.keras.callbacks.EarlyStopping(
            monitor="val_accuracy",
            patience=6,
            restore_best_weights=True,
        ),
        tf.keras.callbacks.ReduceLROnPlateau(
            monitor="val_loss",
            factor=0.5,
            patience=2,
            min_lr=1e-5,
        ),
        tf.keras.callbacks.ModelCheckpoint(
            "resnet18_eurosat_best.h5",
            monitor="val_accuracy",
            save_best_only=True,
        ),
    ]

    resnet18.summary()
```

```python
with tf.device("/CPU:0"):
    history = resnet18.fit(
        train_ds,                    
        validation_data=val_ds,      
        epochs=40,
        callbacks=callbacks,
        class_weight=class_weights,  
    )

```

```python
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report

# 1) test set evaluation
test_loss, test_acc = resnet18.evaluate(test_ds)
print("ResNet18 test loss:", test_loss)
print("ResNet18 test accuracy:", test_acc)

# 2) collect all predictions and true labels
y_true, y_pred = [], []
for x_batch, y_batch in test_ds:
    preds = resnet18.predict(x_batch, verbose=0)
    y_true.extend(y_batch.numpy())
    y_pred.extend(np.argmax(preds, axis=1))

y_true = np.array(y_true)
y_pred = np.array(y_pred)

# 3) print classification report
print(classification_report(y_true, y_pred, target_names=class_names))

# 4) confusion matrix
cm = confusion_matrix(y_true, y_pred)
print("Confusion matrix:\n", cm)

import matplotlib.pyplot as plt

plt.figure(figsize=(7,6))
plt.imshow(cm, interpolation="nearest")
plt.colorbar()
plt.xticks(np.arange(len(class_names)), class_names, rotation=45, ha="right")
plt.yticks(np.arange(len(class_names)), class_names)
plt.xlabel("Predicted label")
plt.ylabel("True label")
plt.title("ResNet18 Confusion Matrix on Test Set")
plt.tight_layout()
plt.show()
```
